```python
import pandas as pd
from statsbombpy import sb
df=sb.events(match_id=3869685)
```

```python
df.columns
```

```python
print(df.columns.tolist())
```

```python
df=df[df['team']=='Argentina']
```

```python
df[['x','y']]=df['location'].apply(pd.Series)
```

```python
df=df[df['pass_outcome'].isna()]
```

```python
df=df[(df['player']=='Lionel Andrés Messi Cuccittini') & (df['type']=='Pass')]
```

```python
df=df[['location','pass_end_location','pass_outcome','player_id']]
```

```python
df.head()
```

```python
df[['x','y']]=df['location'].apply(pd.Series)
```

```python
from mplsoccer import Pitch

pitch = Pitch(pitch_type='statsbomb',pitch_color='white',line_color='black',line_zorder=2)
fig, ax = pitch.draw(figsize=(10, 8))

df_messi=df[df['player_id']==5503]
kde=pitch.kdeplot(df_messi.x,df_messi.y,fill=True,thresh=0,n_levels=100,cut=10,ax=ax,cmap='plasma')
```
