```python
import json
import os
import undetected_chromedriver as uc 
from bs4 import BeautifulSoup
import pandas as pd
```

```python
driver= uc.Chrome()
driver.set_page_load_timeout(5)
```

```python
try:
    driver.get('https://www.whoscored.com/matches/1976987/live/international-fifa-world-cup-2026-republic-of-korea-czechia')
except:
    driver.execute_script('window.stop();')
```

```python
soup= BeautifulSoup(driver.page_source, 'html.parser')
```

```python
data= soup.select_one('script:-soup-contains("matchCentreData")')
```

```python
data.text[:1000]
```

```python
matchdict= json.loads(
    data.text.split("matchCentreData: ")[1].split('matchCentreEventTypeJson')[0].strip()[:-1]
)
```
