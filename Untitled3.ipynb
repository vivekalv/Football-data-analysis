```python
import undetected_chromedriver as uc
from bs4 import BeautifulSoup
import pandas as pd
```

```python
driver= uc.Chrome()
```

```python
driver.get('https://fbref.com/en/comps/Big5/Big-5-European-Leagues-Stats')
```

```python
from io import StringIO
import pandas as pd
import matplotlib.pyplot as plt

df = pd.read_html(StringIO(driver.page_source), 
                  attrs={'id': 'big5_table'})[0]

if isinstance(df.columns, pd.MultiIndex):
    df.columns = df.columns.droplevel(0)

df = df[df['GF'] != 'GF']
df['GF'] = pd.to_numeric(df['GF'], errors='coerce')
df['GA'] = pd.to_numeric(df['GA'], errors='coerce')
df = df.dropna(subset=['GF', 'GA'])
df = df.reset_index(drop=True)

df['difference'] = pd.to_numeric(df['GD'], errors='coerce')
op = df.nlargest(20, 'difference')
up = df.nsmallest(20, 'difference')

# Plot
fig, ax = plt.subplots(figsize=(10, 6))
ax.scatter(df['GF'], df['GA'], alpha=0.7)

ax.xaxis.set_major_locator(plt.MaxNLocator(10))
ax.yaxis.set_major_locator(plt.MaxNLocator(10))

min_val = min(df['GF'].min(), df['GA'].min())
max_val = max(df['GF'].max(), df['GA'].max())
ax.plot([min_val, max_val], [min_val, max_val],
        color='black', linestyle='--', label='GF = GA')

ax.set_xlim(df['GF'].min() - 2, df['GF'].max() + 2)
ax.set_ylim(df['GA'].min() - 2, df['GA'].max() + 2)

for i in range(100):
    ax.text(op.iloc[i]['GF'], op.iloc[i]['GA'], op.iloc[i]['Squad'],
            fontsize=8, fontweight='bold', color='green')
    ax.text(up.iloc[i]['GF'], up.iloc[i]['GA'], up.iloc[i]['Squad'],
            fontsize=8, fontweight='bold', color='red')

ax.set_xlabel('Goals For (GF)', fontsize=12)
ax.set_ylabel('Goals Against (GA)', fontsize=12)
ax.set_title('Goals For vs Goals Against (Big 5 Leagues)', fontsize=14)
ax.legend()

plt.tight_layout()
plt.show()  # ✅ Always LAST
```
