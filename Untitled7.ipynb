```python
import math
import pandas as pd
from statsbombpy import sb
from mplsoccer import VerticalPitch
```

```python
df= sb.events(match_id= 3869685)
```

```python
df[['x','y']]= df['location'].apply(pd.Series)
df=df[df['team']=='Argentina']
```

```python
df['newsecond']=60*df['minute']+df['second']
```

```python
df=df.sort_values(by=['newsecond'])
```

```python
df['newsecond']
```

```python
df['passer']=df['player_id']
df['recipient']=df['passer'].shift(-1)
```

```python
passes=df[df['type']=='Pass']
```

```python
passes['pass_outcome']=passes['pass_outcome'].fillna("Successful")
```

```python
print(passes)
```

```python
completion=passes[passes['pass_outcome']=="Successful"]
```

```python
sub_df=df[df['type']=='Substitution']
first_sub=sub_df['newsecond'].min()
```

```python
if first_sub<=(60*45):
    first_sub=60*45
    
```

```python
completion=completion[completion['newsecond']<first_sub]
```

```python
completion.shape
```

```python
avg_locs_count=completion.groupby('passer').agg({'x':['mean'],'y':['mean','count']})
```

```python
 avg_locs_count.columns=['x','y','count']
```

```python
avg_locs_count.head()
```

```python
passes_between=completion.groupby(['passer','recipient']).id.count().reset_index()
```

```python
passes_between.rename({'id':'pass_count'} ,axis='columns',inplace=True)
```

```python
passes_between=passes_between.merge(avg_locs_count,left_on='passer',right_index=True)
passes_between=passes_between.merge(avg_locs_count,left_on='recipient',right_index=True,suffixes=['','_end'])
```
