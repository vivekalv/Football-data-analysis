```python
import undetected_chromedriver as uc 
import pandas as pd 
from bs4 import BeautifulSoup 
```

```python
driver= uc.Chrome()
```

```python
driver.get('https://fbref.com/en/comps/9/stats/Premier-League-Stats')
```

```python
from io import StringIO

df= pd.read_html(StringIO(driver.page_source), 
    attrs={'id': 'stats_standard'} )[0]
```

```python
df.head()
```

```python
df.columns= ['_'.join(col).strip() for col in df.columns.values]
```

```python
df.head()
```

```python
df.columns
```

```python
import matplotlib.pyplot as plt
df['G+A per 90'] = pd.to_numeric(df['Per 90 Minutes_G+A'], errors='coerce')
df['MP'] = pd.to_numeric(df['Playing Time_MP'], errors='coerce')

best= df.nlargest(10,'G+A per 90')
fig, ax = plt.subplots(figsize=(12, 7))
ax.scatter(best['MP'], best['G+A per 90'], 
           color='steelblue', s=100, alpha=0.8)
for i, row in best.iterrows():
    ax.text(row['MP'], row['G+A per 90'], 
            row['Unnamed: 1_level_0_Player'],  # ← Squad/Player name column
            fontsize=8, fontweight='bold')

ax.set_xlabel('Matches Played', fontsize=12)
ax.set_ylabel('G+A per 90 Minutes', fontsize=12)
ax.set_title('Top 20 Players by G+A per 90 (Min 500 mins)', fontsize=14)

plt.tight_layout()
plt.show()

```

```python
fig, ax = plt.subplots(figsize=(12, 7))

df['G+A per 90'] = pd.to_numeric(df['Per 90 Minutes_G+A'], errors='coerce')
df['Min'] = pd.to_numeric(df['Playing Time_Min'], errors='coerce')
df_filtered = df[df['Min'] >= 500].dropna(subset=['G+A per 90', 'Min'])

# Median lines
x_mid = df_filtered['Min'].median()
y_mid = df_filtered['G+A per 90'].median()

ax.scatter(df_filtered['Min'], df_filtered['G+A per 90'], alpha=0.6)

# Draw quadrant lines
ax.axvline(x_mid, color='gray', linestyle='--', alpha=0.5)
ax.axhline(y_mid, color='gray', linestyle='--', alpha=0.5)

# Label quadrants
ax.text(x_mid + 50, y_mid + 0.1, ' BEST\n(High mins + High G+A)', 
        fontsize=10, color='green', fontweight='bold')
ax.text(x_mid - 800, y_mid + 0.1, 'High G+A\nLow mins', 
        fontsize=9, color='orange')
ax.text(x_mid + 50, y_mid - 0.15, 'High mins\nLow G+A', 
        fontsize=9, color='orange')
ax.text(x_mid - 800, y_mid - 0.15, 'Low mins\nLow G+A', 
        fontsize=9, color='red')

# Label top performers (top right quadrant only)
top = df_filtered[(df_filtered['Min'] > x_mid) & 
                  (df_filtered['G+A per 90'] > y_mid)].nlargest(10, 'G+A per 90')
low= df_filtered[(df_filtered['Min']< x_mid)& (df_filtered['G+A per 90']>y_mid)].nlargest(10,'G+A per 90')

for i, row in top.iterrows():
    ax.text(row['Min'], row['G+A per 90'],
            row['Unnamed: 1_level_0_Player'],
            fontsize=8, fontweight='bold', color='green')
for i, row in low.iterrows():
    ax.text(row['Min'], row['G+A per 90'],
            row['Unnamed: 1_level_0_Player'],
            fontsize=8, fontweight='bold', color='Black')

ax.set_xlabel('Minutes Played', fontsize=12)
ax.set_ylabel('G+A per 90', fontsize=12)
ax.set_title('Best Performers: Minutes vs G+A per 90', fontsize=14)

plt.tight_layout()
plt.show()
```
