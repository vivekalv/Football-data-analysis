```python
import LanusStats as ls

fotmob = ls.FotMob()
```

```python
df=fotmob.get_match_shotmap(4667784)
```

```python
df= df.drop(columns=[x for x in df.columns if 'SCA' in x[0]])
```

```python
df.columns
```

```python
df['situation']
```

```python
df=df['min','playerId', 'playerName','teamName','expectedGoals']
```

```python
# Combine all those columns into one summary column
cols_to_combine = ['isOwnGoal', 'isSavedOffLine', 'isFromInsideBox', 
                   'isBlocked', 'isOnTarget', 'blockedX', 'blockedY', 
                   'goalCrossedY', 'goalCrossedZ']

# Option 1: Combine as a dictionary string (shows keys + values)
df['shot_details'] = df[cols_to_combine].apply(
    lambda row: {col: row[col] for col in cols_to_combine}, axis=1
)

# Option 2: Combine as readable text (cleaner)
df['shot_details'] = df[cols_to_combine].apply(
    lambda row: ', '.join([f"{col}={row[col]}" for col in cols_to_combine]), axis=1
)

# Drop the original columns
df = df.drop(columns=cols_to_combine)

print(df[['playerName', 'teamName', 'expectedGoals', 'shot_details']].head())
```

```python
# Your xG column is called 'expectedGoals' and team is 'teamName'
df['cumulative_xG'] = df.groupby('teamName')['expectedGoals'].cumsum()

print(df[['playerName', 'teamName', 'expectedGoals', 'cumulative_xG']].head(10))
```

```python
df= df.dropna(subset=["min"])
```
